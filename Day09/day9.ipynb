```python
import pandas as pd
```

```python
df = pd.read_csv("../cleaned_dataset.csv")

print("Dataset loaded!")
print(df.head())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum().sum())

df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum().sum())
```

```python
print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates removed!")
print("Remaining duplicates:", df.duplicated().sum())
```

```python
df["category"] = df["category"].str.strip().str.title()
df["segment"] = df["segment"].str.strip().str.title()
df["region"] = df["region"].str.strip().str.title()

print(df[["category", "segment", "region"]].head())
```

```python
df.to_csv("cleaned_day9.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final shape:", df.shape)
```

```python
print("Cleaning Decisions:")

print("1. Removed missing values using dropna().")
print("2. Removed duplicate records using drop_duplicates().")
print("3. Removed extra spaces from text columns.")
print("4. Standardized text using title case.")
print("5. Saved the cleaned dataset as cleaned_day9.csv.")
```
