```python
# DAY 21 - MODEL SELECTION

import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Load Day 20 model evaluation results

metrics_df = pd.read_csv("../Day20/classification_metrics_comparison.csv")

print("Day 20 evaluation results loaded!")
print(metrics_df.round(3))
```

```python
# Select the model with the highest F1 Score

best_model_row = metrics_df.loc[metrics_df["F1 Score"].idxmax()]

print("Best Model:")
print(best_model_row)
```

```python
# Compare all classification models

comparison = metrics_df.sort_values(
    by="F1 Score",
    ascending=False
)

print("Final Model Comparison:")
print(comparison.round(3))
```

```python
# Visual comparison of model F1 Scores

plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Model"],
    comparison["F1 Score"]
)

plt.xlabel("Model")
plt.ylabel("F1 Score")
plt.title("Classification Model Comparison - F1 Score")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig("model_f1_comparison.png", dpi=300)
plt.show()
```

```python
# Final model selection

best_model = comparison.iloc[0]["Model"]
best_f1 = comparison.iloc[0]["F1 Score"]

print("🏆 Best Model:", best_model)
print("F1 Score:", round(best_f1, 3))
```

```python
# Save final model comparison

comparison.to_csv("final_model_comparison.csv", index=False)

print("Final model comparison saved successfully!")
```

```python
# Model selection justification

print("Model Selection Justification")
print()
print(f"We selected {best_model} as the best model based on the highest F1 Score.")
print()
print("F1 Score is useful because it balances Precision and Recall.")
print("This is important for fraud detection because both false alarms")
print("and missed fraud cases can have business impact.")
print()
print("The final model should also be considered for interpretability,")
print("reliability, scalability, and real-world business requirements.")
```
