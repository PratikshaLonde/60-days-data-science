```python
df = pd.read_csv("../cleaned_dataset.csv")

df["order_date"] = pd.to_datetime(df["order_date"])

print("Main dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
monthly_revenue = (
    df.groupby(df["order_date"].dt.to_period("M"))["sales"]
    .sum()
    .reset_index()
)

monthly_revenue["order_date"] = (
    monthly_revenue["order_date"].astype(str)
)

plt.figure(figsize=(11, 5))

plt.plot(
    monthly_revenue["order_date"],
    monthly_revenue["sales"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Revenue trend created successfully!")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(
    monthly_revenue["order_date"],
    monthly_revenue["sales"],
    marker="o"
)

axes[0].set_title("Monthly Revenue")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Revenue")
axes[0].tick_params(axis="x", rotation=45)

axes[1].bar(
    risk_counts.index,
    risk_counts.values
)

axes[1].set_title("Customer Risk Distribution")
axes[1].set_xlabel("Risk Level")
axes[1].set_ylabel("Customers")

plt.tight_layout()
plt.show()

print("Unified executive dashboard created successfully!")
```

```python
customer_orders = (
    df.groupby("customer_name")["order_id"]
    .nunique()
)

repeat_customers = (
    customer_orders[customer_orders > 1].count()
)

total_customers = customer_orders.count()

repeat_customer_rate = (
    repeat_customers / total_customers
) * 100

retention_summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Repeat Customers",
        "Repeat Customer Rate"
    ],
    "Value": [
        total_customers,
        repeat_customers,
        round(repeat_customer_rate, 2)
    ]
})

print("Customer retention analysis integrated successfully!")
display(retention_summary)
```

```python
# Load Day 41 risk data again
risk_data = pd.read_csv("../Day41/customer_risk_ranking.csv")

# Calculate customer risk counts
risk_counts = (
    risk_data["risk_level"]
    .value_counts()
    .reindex(
        ["Low Risk", "Medium Risk", "High Risk"],
        fill_value=0
    )
)

low_risk_customers = int(risk_counts["Low Risk"])
medium_risk_customers = int(risk_counts["Medium Risk"])
high_risk_customers = int(risk_counts["High Risk"])

average_risk = risk_data["risk_percentage"].mean()

# Calculate retention
customer_orders = (
    df.groupby("customer_name")["order_id"]
    .nunique()
)

total_customers = customer_orders.count()

repeat_customers = (
    customer_orders[customer_orders > 1].count()
)

repeat_customer_rate = (
    repeat_customers / total_customers
) * 100

# Create unified executive dashboard
executive_dashboard = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Profit",
        "Total Customers",
        "Repeat Customer Rate",
        "High-Risk Customers",
        "Medium-Risk Customers",
        "Low-Risk Customers",
        "Average Churn Risk"
    ],
    "Value": [
        round(df["sales"].sum(), 2),
        round(df["profit"].sum(), 2),
        total_customers,
        round(repeat_customer_rate, 2),
        high_risk_customers,
        medium_risk_customers,
        low_risk_customers,
        round(average_risk, 2)
    ]
})

print("Unified executive KPI dashboard created successfully!")

display(executive_dashboard)
```

```python
kpi_chart_data = executive_dashboard[
    executive_dashboard["KPI"].isin([
        "Total Customers",
        "High-Risk Customers",
        "Medium-Risk Customers",
        "Low-Risk Customers"
    ])
]

plt.figure(figsize=(10, 5))

plt.bar(
    kpi_chart_data["KPI"],
    kpi_chart_data["Value"]
)

plt.title("Executive Customer Intelligence Overview")
plt.xlabel("Business KPI")
plt.ylabel("Count")

plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

print("Executive KPI visualization created successfully!")
```

```python
forecast_data = pd.read_csv("../Day37/revenue_30_day_forecast.csv")

print("Day 37 revenue forecast loaded successfully!")
print("Columns:")
print(forecast_data.columns.tolist())

print("\nFirst 5 rows:")
display(forecast_data.head())
```

```python
forecast_data["date"] = pd.to_datetime(forecast_data["date"])

plt.figure(figsize=(12, 5))

plt.plot(
    forecast_data["date"],
    forecast_data["predicted_revenue"],
    marker="o"
)

plt.title("30-Day Revenue Forecast")
plt.xlabel("Date")
plt.ylabel("Predicted Revenue")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Revenue forecasting visualization created successfully!")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Revenue Forecast
axes[0].plot(
    forecast_data["date"],
    forecast_data["predicted_revenue"],
    marker="o"
)
axes[0].set_title("30-Day Revenue Forecast")
axes[0].set_xlabel("Date")
axes[0].set_ylabel("Predicted Revenue")
axes[0].tick_params(axis="x", rotation=45)

# Customer Risk
axes[1].bar(
    risk_counts.index,
    risk_counts.values
)
axes[1].set_title("Customer Risk Distribution")
axes[1].set_xlabel("Risk Level")
axes[1].set_ylabel("Customers")
axes[1].tick_params(axis="x", rotation=20)

# Customer Retention
axes[2].bar(
    ["Repeat Customers", "Total Customers"],
    [repeat_customers, total_customers]
)
axes[2].set_title("Customer Retention Overview")
axes[2].set_ylabel("Customers")

plt.tight_layout()
plt.show()

print("Unified business intelligence dashboard created successfully!")
```
